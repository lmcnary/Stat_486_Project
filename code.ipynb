```python
import pandas as pd

df = pd.read_csv("data/car_pricing_final.csv")

print("Duplicates:", df.duplicated().sum())

print(df["type"].value_counts())

print("Missing MSRP:", df["Trim MSRP"].isna().sum())
print("Missing gasoline MPG:",
      df["Mileage Combined Mpg"].isna().sum())

print(df.groupby("type")["10yr_cost"].describe())
```
