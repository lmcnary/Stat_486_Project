```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

from sklearn.compose import ColumnTransformer

from sklearn.metrics import mean_absolute_error, r2_score
```

```python
df = pd.read_csv("data/car_pricing_final.csv")
gas_df = df[df["type"] == "gas"]
gas_df.head()


X_gas = gas_df[["doors", "body type", "engine liters", "cylinders", "turbo", "transmission"]]
y_gas = gas_df["mileage"]
```

```python
hybrid_df = df[df["type"] == "hybrid"]
hybrid_df.head()
X_hybrid = hybrid_df[["doors", "body type", "engine liters", "cylinders", "turbo", "transmission"]]
y_hybrid = hybrid_df["mileage"]
```

```python
#Preprocessing
X_train_gas, X_test_gas, y_train_gas, y_test_gas = train_test_split(X_gas, y_gas, test_size=0.2, random_state=42)

numeric_features = ["doors", "engine liters", "cylinders",]
categorical_features = ["body type", "transmission"]

numeric_pre = Pipeline([
    ("imputer", SimpleImputer(strategy= "median"))
])
categorical_pre = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
preprocessor = ColumnTransformer([
    ("numeric", numeric_pre, numeric_features),
    ("categorical", categorical_pre, categorical_features)
])
```

```python
#Pipeline
for depth in [2, 5, 10, 20]:
    for leaf in [5, 10, 20, 50]:
        dt = Pipeline([
            ("preprocessor", preprocessor),
            ("model", DecisionTreeRegressor(max_depth=depth, min_samples_leaf=leaf, random_state=42))
        ])
        dt.fit(X_train_gas, y_train_gas)
        y_pred = dt.predict(X_test_gas)
        mae = mean_absolute_error(y_test_gas, y_pred)
        train_accuracy = dt.score(X_train_gas, y_train_gas)
        test_accuracy = dt.score(X_test_gas, y_test_gas)
        r2 = r2_score(y_test_gas, y_pred)

        print(f"Depth: {depth}, Leaf: {leaf}, MAE: {mae}, R2: {r2}, Train Accuracy: {train_accuracy}, Test Accuracy: {test_accuracy}")
```
