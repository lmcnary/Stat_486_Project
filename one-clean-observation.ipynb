```python
import pandas as pd

df = pd.read_csv("data/car_pricing_final.csv")

print("Duplicates:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())
print("missing mileage values:", df["mileage"].isnull().sum())
print()
print(df["type"].value_counts())
print()
print("mileage summary by type")
print(df.groupby("type")["mileage"].describe())
```

We evaluated our clean data using  using duplicate checks, missingness checks, fuel type counts, and descriptive statistics. The data set contains 11,854 gasoline vehicles and 807 hybrids. no exact duplicate rows were identified. There were 194 missing values across the dataset, but no missing mileage observations. Missing values in vehicle characteristics were handled using SimpleImputer(). Specifically for numeric variables median imputation and for categorical variables most-frequent imputation. On average gasoline vehicles have a mean combined mfuel efficiency of 22.76 mpg and hybrids averaged 34.28 mpg. a random sample of ten cleaned observations is provided below to demonstrate the dataset's structure and consistency for every one clean observation.
